```python
class student:
    def __init__(self, name, age):
        self._name = name
        self.age = age
        #s1 print address of the object
        print(f"Address of the object: {id(self)}")
    def __str__(self):
        return f"Student(name={self.name}, age={self.age})"
     #print (s1) will call __str__ method or str(s1)
    @property
    def name(self):
        return self._name 
    def __repr__(self):
        return f"Student(name={self.name}, age={self.age})"
    #s1 will print the string representation of the object
    def display(self):
        print(f"Name: {self.name}, Age: {self.age}")
```

```python
s1 = student("John", 20)
print(s1)  # This will call the __str__ method
print(repr(s1))  # This will call the __repr__ method
str(s1)  # This will also call the __str__ method
s1  # This will call the __repr__ method in the interactive environment
```
