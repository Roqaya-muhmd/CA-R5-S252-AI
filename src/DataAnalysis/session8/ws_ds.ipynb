```python
def replace_outliers (df):
    for col in df.select_dtypes("number").columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        # print(Q1, Q3)
        IQR = Q3 - Q1
        lower_fence = Q1 - 1.5 * IQR
        upper_fence = Q3 + 1.5 * IQR
        df[col] = df[col].clip(lower_fence, upper_fence)
    return df
```
