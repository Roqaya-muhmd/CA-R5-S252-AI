Class defination

```python
class Student:
    name = ""
    age = 0
    grade = 0
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def get_grade(self):
        return self.grade
```

```python
s1 = Student("John", 15, 90)
s1.name="Roqaya"
print(s1.name)
s1.get_grade()
```
