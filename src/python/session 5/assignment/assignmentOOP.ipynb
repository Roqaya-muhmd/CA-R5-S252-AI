```python
class twin_primes:
    def __init__(self):
        self.primes = []
        self.twin_prime=[]
        self.last_prime = 3
        self.find_twin_primes() 
    def find_twin_primes(self):
        for i in range(2, 1000):
            is_prime = True
            for j in range(2, i):
                if i % j == 0:
                    is_prime = False
                    break
            if is_prime:
                self.primes.append(i)
                if self.last_prime == i - 2:
                    self.twin_prime.append((self.last_prime, i))
                self.last_prime = i

    def get_primes(self):
        return self.primes

    def get_twin_primes(self):
        return self.twin_prime

            
    
```

```python
class factorization:
    def __init__(self, num):
        self.num=num
        self.primes = twin_primes().get_primes()
    def factorize(self):
        factors=[]
        for i in self.primes:
            while self.num%i==0:
                self.num=self.num//i
                factors.append(i)
        
        return factors
```

```python
class decimal_to_binary:
    def __init__(self, num):
        self.num=num
    def decTobin(self):
        bin:str=""
        while self.num//2!=0 or self.num%2!=0:
            bin+=str(self.num%2)
            self.num=self.num//2
        
        return bin[::-1]
```

```python
print(decimal_to_binary(11).decTobin())
```

```python
class perfect_numbers:
    def __init__(self, l, r):
        self.l=l
        self.r=r
    def perfect_nums(self):
        perfects=[]
        for i in range(self.l,self.r):
            total=0
            for j in range(1,i):
                if i%j==0:
                    total+=j
            if total==i :
                perfects.append(i)
        return perfects
    
```

```python
perfect_numbers(0,100).perfect_nums()
```
