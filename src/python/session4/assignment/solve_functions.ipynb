```python
def perfect_nums(l,r):
    perfects=[]
    for i in range(l,r):
        total=0
        for j in range(1,i):
            if i%j==0:
                total+=j
        if total==i :
            perfects.append(i)
    return perfects
```

```python
perfect_nums(0,100)
```
