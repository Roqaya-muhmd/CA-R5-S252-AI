```python
def email_info(string:str):
    if not check_invalidation(string):
        return "Invalid email address"
    extracted_username=extract_username(string)
    extracted_domain=extract_domain(string)
    return f"Username: {extracted_username}, Domain: {extracted_domain}, Type: {check_domain(string)}"
    
  

    
```

```python
def check_invalidation(string:str):
    if string.count('@') != 1:
        return False
    
    if string.count('.') == 0:
        return False
    
    at_index=string.index('@')
    dot_index=string.index('.')
    
    if at_index == 0 or dot_index == len(string) - 1:
        return False
    
    if dot_index < at_index:
        return False
    
    return True
```

```python
def extract_username(string:str):
    if not check_invalidation(string):
        return "Invalid email address"
    
    at_index=string.index('@')
    return string[:at_index]
```

```python
def extract_domain(string:str):
    if not check_invalidation(string):
        return "Invalid email address"
    
    dot_index=string.index('.')
    return string[dot_index+1:]
```

```python
def check_domain(string:str):
    if not check_invalidation(string):
        return "Invalid email address"
    
    domain=extract_domain(string)
    if domain=='com':
        return "Commercial"
    
    elif domain=='edu':
        return "Educational"
    else:
        return "Other"
```
