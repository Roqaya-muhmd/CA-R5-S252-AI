```python
def extract_reverse(text: str) -> str:
    """
    Extracts the reverse of the word after @ character.

    Args:
        text (str): The input string to be reversed.

    Returns:
        str: The reversed string.
    """
    words_2 = text.split()
    word1_2 = words_2[0][::-1]
    word2_2 = words_2[1]
    decoded_2 = f"{word1_2} {word2_2}"
    return decoded_2
```

```python
example_text = input("Enter a string with @ character: ")
reversed_word = extract_reverse(example_text)
print(f"Reversed word: {reversed_word}")
```
